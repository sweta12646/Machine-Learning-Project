### **Gradient Boosting**

```python
X = df[
    [
        "quantity",
        "unit_price",
        "discount",
        "delivery_days",
        "customer_rating"
    ]
]

Y = df[["revenue"]]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr = GradientBoostingRegressor(
    n_estimators=100,      # number of trees
    learning_rate=0.1,     # contribution of each tree
    max_depth=3,           # complexity of each tree
    random_state=42
)
```

```python
gbr.fit(X_train, Y_train.values.ravel())
```

```python
Y_pred = gbr.predict(X_test)

Y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(
    Y_test,
    Y_pred
)

mse = mean_squared_error(
    Y_test,
    Y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    Y_test,
    Y_pred
)
```

```python
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 score:", round(r2, 2))
```

```python
Y_train_pred = gbr.predict(X_train)

training_r2 = r2_score(
    Y_train,
    Y_train_pred
)

print(
    "Training Accuracy:",
    round(training_r2 * 100, 2),
    "%"
)
```

```python
test_r2 = r2_score(
    Y_test,
    Y_pred
)

print(
    "Test Accuracy / Model Efficiency:",
    round(test_r2 * 100, 2),
    "%"
)
```

```python
plt.figure(figsize=(6.5,6))

plt.scatter(
    Y_test,
    Y_pred,
    color="green",
    label="Predicted Vs Actual"
)

plt.plot(
    [Y_test.min(), Y_test.max()],
    [Y_test.min(), Y_test.max()],
    color="red",
    linestyle="--",
    label="perfect prediction line"
)

plt.xlabel("Actual Revenue")
plt.ylabel("Predicted Revenue")

plt.title(
    "Gradient Boosting Regression - Actual VS Prediction"
)

plt.legend()

plt.show()
```

### **Overall comparison**

```python
# Overall Model Comparison

models = [
    "Simple Linear Regression",
    "Multiple Linear Regression",
    "Gradient Boosting Regression"
]

training_accuracy = [
    38.06,
    87.10,
    99.85
]

test_accuracy = [
    41.05,
    86.07,
    99.80
]

for i in range(len(models)):
    print(models[i])
    print("Training Accuracy:", training_accuracy[i], "%")
    print("Test Accuracy / Model Efficiency:", test_accuracy[i], "%")
    print("-" * 50)
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = [
    "Simple Linear",
    "Multiple Linear",
    "Gradient Boosting"
]

training_accuracy = [38.06, 87.10, 99.85]
test_accuracy = [41.05, 86.07, 99.80]

x = np.arange(len(models))
width = 0.35

plt.figure(figsize=(9, 6))

plt.bar(
    x - width/2,
    training_accuracy,
    width,
    label="Training Accuracy"
)

plt.bar(
    x + width/2,
    test_accuracy,
    width,
    label="Test Accuracy"
)

plt.xlabel("Regression Models")
plt.ylabel("Accuracy (%)")
plt.title("Overall Model Accuracy Comparison")

plt.xticks(x, models)
plt.ylim(0, 110)

plt.legend()
plt.show()
```


Overall Model Comparison And their Insight:

1. Simple Linear Regression:
   Training Accuracy = 38.06%
   Test Accuracy = 41.05%

   only quantity to predict revenue.

   Training Accuracy = 87.10%

   The model performs much better because it uses multiple

   and customer rating.

3. Gradient Boosting Regression:
   Training Accuracy = 99.85%

   The model provides the best performance among all three models.
   The actual vs predicted graph also shows predictions very close

Conclusion:
Gradient Boosting Regression is the best-performing model,

has the lowest performance.
